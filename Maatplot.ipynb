```python
# compar oscar vs non oscar
import numpy as  np
import matplotlib.pyplot as plt
oscar_movies=[
    "The Dark night",
    "The Hurt Locker",
    "The King Speech",
    "The Artist"
     "Game Of Throns"
]
years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,178,427,133,150]

non_oscar_movies=["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoin"]
non_oscar_Revenue=[378,2788,829,185,275]

x=np.arange(len(years))
# plot
width=0.4
plt.bar(x-width/2,oscar_revenue,width,label="Oscar_Revenue") #(x-width/2)=shifting left
plt.bar(x+width/2,non_oscar_Revenue,width,label="No_oscar_Revenue") #(x+width/2)shifting right

plt.title("Oscar_Movie Vs Non_oscar_Movie")

plt.xlabel("Movies")
plt.ylabel("Revenue (in $M)")
plt.xticks(x,years) #converted into array index 0,1,2,3 to year


```

```python
#scatter plot

people=["Person A","Person B","Person c","Person D","Person E",
        "Person F","Person G","Person H","Person I","Person J"]
age=[22,25,30,35,40,45,50,55,60,65]
bp=[110,115,120,122,125,138,135,123,145,158]

# colors=["red","purple","yellow","blue","pink","red","purple","yellow","blue","pink"]
# plt.scatter(age,blood_pressure,color=colors)

# apply the colr condition
# colors=["green" if x <135 else "red" for x in bp]
plt.scatter(age,bp,s=bp,cmap="OrRd",c=bp)
#s=bp sets the size of each point according to blood pressure.
# c=bp colors the points according to blood pressure values.
# cmap="OrRd" applies an orange-red color gradient.

# annotations
for i in range(len(people)):
  plt.annotate(people[i],xy=(age[i],bp[i]),xytext=(age[i]+3,bp[i]+1))

plt.xlim(min(age),max(age)+15)
plt.ylim(min(bp),max(bp)+5)
plt.title("Age VS Bp")
plt.xlabel("Age")
plt.grid()
plt.ylabel("Blood Pressure")
plt.colorbar()
```

```python
#pie charts
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[588,150,120,100,50]
colors=["red","black","pink","blue","yellow"]

explode=[0,0,0,0.2,0]
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
        wedgeprops={
            "edgecolor":"black",
            "linewidth":1,
            "linestyle":"--"
        },
        explode=explode,
        startangle=60
)
plt.style.available
# explode separates a slice from the pie chart. Here, the fourth slice moves outward by 0.2.

# wedgeprops customizes the borders of the pie slices.
# - edgecolor="black" — black borders.
# - linewidth=1 — border thickness.
# - linestyle="--" — dashed borders.
```
